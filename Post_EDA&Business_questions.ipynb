```python
categorical_columns = df.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

boolean_columns = df.select_dtypes(
    include=["bool", "boolean"]
).columns.tolist()

categorical_columns = list(
    dict.fromkeys(categorical_columns + boolean_columns)
)

category_summaries = {}

for column in categorical_columns:

    summary = (
        df[column]
        .value_counts(dropna=False)
        .head(20)
        .rename_axis("value")
        .reset_index(name="records")
    )

    summary["share_pct"] = summary["records"] / len(df) * 100

    category_summaries[column] = summary

for column in [
    "borough",
    "incident_dispatch_area",
    "initial_call_type",
    "final_call_type",
    "initial_severity_level",
    "final_severity_level",
    "incident_disposition_code"
]:

    if column in category_summaries:
        print(f"\nTop values — {column}")
        display(category_summaries[column])

category_export = []

for column, summary in category_summaries.items():

    temp = summary.copy()
    temp.insert(0, "column", column)
    category_export.append(temp)

if category_export:

    pd.concat(
        category_export,
        ignore_index=True
    ).to_csv(
        EDA_OUTPUT_DIR / "categorical_value_summaries.csv",
        index=False
    )
```

```python
def response_metric_summary(seconds_column, validity_column, metric_name):

    valid_mask = df[validity_column].eq(True)

    values_seconds = pd.to_numeric(
        df.loc[valid_mask, seconds_column],
        errors="coerce"
    ).dropna()

    values_minutes = values_seconds / 60

    return {
        "metric": metric_name,
        "all_incidents": len(df),
        "valid_flag_incidents": int(valid_mask.sum()),
        "measured_values": int(values_minutes.count()),
        "measurement_coverage_pct": (
            values_minutes.count() / len(df) * 100
        ),
        "mean_minutes": values_minutes.mean(),
        "median_minutes": values_minutes.median(),
        "p90_minutes": values_minutes.quantile(0.90),
        "p95_minutes": values_minutes.quantile(0.95),
        "zero_second_count": int(values_seconds.eq(0).sum()),
        "zero_pct_of_measured": (
            values_seconds.eq(0).mean() * 100
            if len(values_seconds) else np.nan
        )
    }


response_summary = pd.DataFrame([

    response_metric_summary(
        "dispatch_response_seconds",
        "is_valid_dispatch_response",
        "Dispatch response"
    ),

    response_metric_summary(
        "incident_response_seconds",
        "is_valid_incident_response",
        "Incident response"
    )

])

display(response_summary.round(2))

response_summary.to_csv(
    EDA_OUTPUT_DIR / "response_time_summary.csv",
    index=False
)
```

```python

# ============================================================
# CELL 7 — ANSWER ALL 16 BUSINESS QUESTIONS
# ============================================================

# Helper function: save each result as a CSV file
def save_result(data, filename):
    data.to_csv(EDA_OUTPUT_DIR / filename, index=False)
    print(f"Saved: {filename}")


# Helper function: prepare datetime columns safely
def get_datetime(column):
    return pd.to_datetime(df[column], errors="coerce")


# Helper function: calculate response-time statistics
def response_statistics(data, group_column, seconds_column, validity_column):

    temp = data[[group_column, seconds_column, validity_column]].copy()

    temp = temp[temp[validity_column].eq(True)]

    temp[seconds_column] = pd.to_numeric(
        temp[seconds_column], errors="coerce"
    )

    temp = temp.dropna(subset=[seconds_column, group_column])

    temp["minutes"] = temp[seconds_column] / 60

    result = (
        temp.groupby(group_column, dropna=False)["minutes"]
        .agg(
            valid_measurements="count",
            mean_minutes="mean",
            median_minutes="median",
            p90_minutes=lambda x: x.quantile(0.90),
            p95_minutes=lambda x: x.quantile(0.95)
        )
        .reset_index()
    )

    return result


# ------------------------------------------------------------
# BQ01 — How many incidents were recorded in each month/year?
# ------------------------------------------------------------

incident_dates = get_datetime("incident_datetime")

bq01 = pd.DataFrame({
    "incident_month": incident_dates.dt.to_period("M").astype(str)
})

bq01 = (
    bq01.groupby("incident_month")
    .size()
    .reset_index(name="incident_count")
    .sort_values("incident_month")
)

display(bq01)
save_result(bq01, "BQ01_monthly_incident_volume.csv")


# ------------------------------------------------------------
# BQ02 — Which borough has the highest incident volume?
# ------------------------------------------------------------

bq02 = (
    df.groupby("borough", dropna=False)
    .size()
    .reset_index(name="incident_count")
    .sort_values("incident_count", ascending=False)
)

bq02["share_pct"] = bq02["incident_count"] / len(df) * 100

display(bq02)
save_result(bq02, "BQ02_incidents_by_borough.csv")


# ------------------------------------------------------------
# BQ03 — Which dispatch areas receive the most incidents?
# ------------------------------------------------------------

bq03 = (
    df.groupby("incident_dispatch_area", dropna=False)
    .size()
    .reset_index(name="incident_count")
    .sort_values("incident_count", ascending=False)
)

bq03["share_pct"] = bq03["incident_count"] / len(df) * 100

display(bq03)
save_result(bq03, "BQ03_incidents_by_dispatch_area.csv")


# ------------------------------------------------------------
# BQ04 — What are the most common initial call types?
# ------------------------------------------------------------

bq04 = (
    df.groupby("initial_call_type", dropna=False)
    .size()
    .reset_index(name="incident_count")
    .sort_values("incident_count", ascending=False)
)

bq04["share_pct"] = bq04["incident_count"] / len(df) * 100

display(bq04.head(20))
save_result(bq04, "BQ04_initial_call_types.csv")


# ------------------------------------------------------------
# BQ05 — Weekday vs weekend incident volume
# ------------------------------------------------------------

temp = pd.DataFrame({
    "incident_date": incident_dates.dt.date,
    "day_number": incident_dates.dt.dayofweek
})

temp = temp.dropna(subset=["incident_date"])

temp["day_type"] = np.where(
    temp["day_number"] >= 5,
    "Weekend",
    "Weekday"
)

# Count incidents for each calendar date
daily_counts = (
    temp.groupby(["incident_date", "day_type"])
    .size()
    .reset_index(name="incident_count")
)

bq05 = (
    daily_counts.groupby("day_type")
    .agg(
        total_incidents=("incident_count", "sum"),
        number_of_days=("incident_date", "nunique"),
        average_incidents_per_day=("incident_count", "mean")
    )
    .reset_index()
)

display(bq05)
save_result(bq05, "BQ05_weekday_vs_weekend.csv")


# ------------------------------------------------------------
# BQ06 — Dispatch response time statistics
# ------------------------------------------------------------

bq06 = response_statistics(
    df.assign(response_group="All incidents"),
    "response_group",
    "dispatch_response_seconds",
    "is_valid_dispatch_response"
)

display(bq06)
save_result(bq06, "BQ06_dispatch_response_statistics.csv")


# ------------------------------------------------------------
# BQ07 — Incident response time statistics
# ------------------------------------------------------------

bq07 = response_statistics(
    df.assign(response_group="All incidents"),
    "response_group",
    "incident_response_seconds",
    "is_valid_incident_response"
)

display(bq07)
save_result(bq07, "BQ07_incident_response_statistics.csv")


# ------------------------------------------------------------
# BQ08 — Which borough has the longest median response time?
# ------------------------------------------------------------

bq08 = response_statistics(
    df,
    "borough",
    "incident_response_seconds",
    "is_valid_incident_response"
)

bq08 = bq08.sort_values("median_minutes", ascending=False)

display(bq08)
save_result(bq08, "BQ08_response_time_by_borough.csv")


# ------------------------------------------------------------
# BQ09 — Do response times differ by initial call type?
# ------------------------------------------------------------

bq09 = response_statistics(
    df,
    "initial_call_type",
    "incident_response_seconds",
    "is_valid_incident_response"
)

bq09 = bq09.sort_values("median_minutes", ascending=False)

display(bq09)
save_result(bq09, "BQ09_response_time_by_call_type.csv")


# ------------------------------------------------------------
# BQ10 — How often does final severity differ from initial?
# ------------------------------------------------------------

initial_severity = df["initial_severity_level"]
final_severity = df["final_severity_level"]

both_severity_present = (
    initial_severity.notna() & final_severity.notna()
)

severity_changed = (
    initial_severity.astype("string")
    != final_severity.astype("string")
)

comparable_count = int(both_severity_present.sum())
changed_count = int(
    (severity_changed & both_severity_present).sum()
)

bq10 = pd.DataFrame({
    "comparable_incidents": [comparable_count],
    "changed_severity_incidents": [changed_count],
    "unchanged_severity_incidents": [
        comparable_count - changed_count
    ],
    "changed_percentage": [
        changed_count / comparable_count * 100
        if comparable_count else np.nan
    ]
})

print("Severity comparison summary:")
display(bq10)

print("Initial vs final severity cross-tabulation:")

severity_crosstab = pd.crosstab(
    initial_severity,
    final_severity,
    margins=True
)

display(severity_crosstab)

save_result(bq10, "BQ10_severity_change_summary.csv")

severity_crosstab.to_csv(
    EDA_OUTPUT_DIR / "BQ10_severity_crosstab.csv"
)


# ------------------------------------------------------------
# BQ11 — Response times across initial severity levels
# ------------------------------------------------------------

bq11 = response_statistics(
    df,
    "initial_severity_level",
    "incident_response_seconds",
    "is_valid_incident_response"
)

display(bq11)
save_result(bq11, "BQ11_response_by_initial_severity.csv")


# ------------------------------------------------------------
# BQ12 — Percentage of incidents with operational flags
# ------------------------------------------------------------

operational_flags = [
    "is_held",
    "is_reopened",
    "is_standby",
    "is_transfer",
    "is_special_event"
]

bq12_rows = []

for column in operational_flags:

    if column not in df.columns:
        continue

    true_count = int(df[column].eq(True).sum())
    known_count = int(df[column].notna().sum())

    bq12_rows.append({
        "operational_flag": column,
        "true_incidents": true_count,
        "known_values": known_count,
        "missing_values": int(df[column].isna().sum()),
        "percentage_of_all_incidents": true_count / len(df) * 100,
        "percentage_of_known_values": (
            true_count / known_count * 100
            if known_count else np.nan
        )
    })

bq12 = pd.DataFrame(bq12_rows)

display(bq12)
save_result(bq12, "BQ12_operational_flags.csv")


# ------------------------------------------------------------
# BQ13 — Which disposition codes occur most frequently?
# ------------------------------------------------------------

bq13 = (
    df.groupby("incident_disposition_code", dropna=False)
    .size()
    .reset_index(name="incident_count")
    .sort_values("incident_count", ascending=False)
)

bq13["share_pct"] = bq13["incident_count"] / len(df) * 100

display(bq13.head(25))
save_result(bq13, "BQ13_disposition_codes.csv")


# ------------------------------------------------------------
# BQ14 — How does response-time validity vary over time?
# ------------------------------------------------------------

bq14_temp = pd.DataFrame({
    "month": incident_dates.dt.to_period("M").astype(str),
    "dispatch_valid": df["is_valid_dispatch_response"].eq(True),
    "incident_valid": df["is_valid_incident_response"].eq(True)
})

bq14 = (
    bq14_temp.groupby("month")
    .agg(
        total_incidents=("month", "size"),
        valid_dispatch_count=("dispatch_valid", "sum"),
        valid_incident_count=("incident_valid", "sum")
    )
    .reset_index()
)

bq14["dispatch_valid_pct"] = (
    bq14["valid_dispatch_count"] / bq14["total_incidents"] * 100
)

bq14["incident_valid_pct"] = (
    bq14["valid_incident_count"] / bq14["total_incidents"] * 100
)

display(bq14)
save_result(bq14, "BQ14_response_validity_by_month.csv")


# ------------------------------------------------------------
# BQ15 — Geographic field completeness across boroughs
# ------------------------------------------------------------

geographic_columns = [
    "zipcode",
    "police_precinct",
    "community_district"
]

bq15_rows = []

for borough, group in df.groupby("borough", dropna=False):

    for column in geographic_columns:

        if column not in df.columns:
            continue

        total = len(group)
        missing = int(group[column].isna().sum())

        bq15_rows.append({
            "borough": borough,
            "geographic_field": column,
            "total_incidents": total,
            "available_values": total - missing,
            "missing_values": missing,
            "missing_percentage": (
                missing / total * 100 if total else np.nan
            ),
            "completeness_percentage": (
                (total - missing) / total * 100
                if total else np.nan
            )
        })

bq15 = pd.DataFrame(bq15_rows)

display(bq15)
save_result(bq15, "BQ15_geographic_completeness.csv")


# ------------------------------------------------------------
# BQ16 — How often are hospital timestamps recorded together?
# ------------------------------------------------------------

hospital_columns = [
    "first_to_hosp_datetime",
    "first_hosp_arrival_datetime"
]

hospital_temp = df[hospital_columns].copy()

hospital_temp["departure_present"] = (
    hospital_temp["first_to_hosp_datetime"].notna()
)

hospital_temp["arrival_present"] = (
    hospital_temp["first_hosp_arrival_datetime"].notna()
)

conditions = [
    hospital_temp["departure_present"] & hospital_temp["arrival_present"],
    hospital_temp["departure_present"] & ~hospital_temp["arrival_present"],
    ~hospital_temp["departure_present"] & hospital_temp["arrival_present"]
]

choices = [
    "Both timestamps recorded",
    "Departure only",
    "Arrival only"
]

hospital_temp["timestamp_status"] = np.select(
    conditions,
    choices,
    default="Neither timestamp recorded"
)

bq16 = (
    hospital_temp["timestamp_status"]
    .value_counts()
    .rename_axis("timestamp_status")
    .reset_index(name="incident_count")
)

bq16["percentage"] = bq16["incident_count"] / len(df) * 100

display(bq16)
save_result(bq16, "BQ16_hospital_timestamp_completeness.csv")


# ============================================================
# FINAL CHECK — ALL BUSINESS QUESTIONS
# ============================================================

print("\nALL 16 BUSINESS QUESTIONS HAVE BEEN ANALYZED.")

for i in range(1, 17):
    print(f"BQ{i:02d}: Result generated")

print("\nResults folder:")
print(EDA_OUTPUT_DIR)
```
